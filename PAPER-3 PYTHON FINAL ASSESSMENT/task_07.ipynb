```python
from abc import ABC, abstractmethod


class Employee(ABC):
    def __init__(self, emp_id, name):
        if not emp_id or type(emp_id) != str:
            raise ValueError("Employee ID valid string honi chahiye")
        if not name or type(name) != str:
            raise ValueError("Employee name valid string honi chahiye")

        self.emp_id = emp_id
        self.name = name

    @abstractmethod
    def calculate_pay(self):
        pass

    def get_role(self):
        return self.__class__.__name__


class SalariedEmployee(Employee):
    def __init__(self, emp_id, name, salary):
        super().__init__(emp_id, name)
        if type(salary) not in (int, float) or salary < 0:
            raise ValueError("Salary negative nahi ho sakti")
        self.salary = float(salary)

    def calculate_pay(self):
        return round(self.salary, 2)


class HourlyEmployee(Employee):
    def __init__(self, emp_id, name, hourly_rate, hours_worked):
        super().__init__(emp_id, name)
        if type(hourly_rate) not in (int, float) or hourly_rate < 0:
            raise ValueError("Hourly rate negative nahi ho sakta")
        if type(hours_worked) not in (int, float) or hours_worked < 0:
            raise ValueError("Hours worked negative nahi ho sakte")

        self.hourly_rate = float(hourly_rate)
        self.hours_worked = float(hours_worked)

    def calculate_pay(self):
        if self.hours_worked <= 40:
            pay = self.hours_worked * self.hourly_rate
        else:
            regular_pay = 40 * self.hourly_rate
            overtime_hours = self.hours_worked - 40
            overtime_pay = overtime_hours * (self.hourly_rate * 1.5)
            pay = regular_pay + overtime_pay
        return round(pay, 2)


class CommissionEmployee(Employee):
    def __init__(self, emp_id, name, base_salary, sales, commission_rate):
        super().__init__(emp_id, name)
        if type(base_salary) not in (int, float) or base_salary < 0:
            raise ValueError("Base salary negative nahi ho sakti")
        if type(sales) not in (int, float) or sales < 0:
            raise ValueError("Sales negative nahi ho sakti")
        if type(commission_rate) not in (int, float) or not (0 <= commission_rate <= 1):
            raise ValueError("Commission rate 0 aur 1 ke darmiyan hona chahiye")

        self.base_salary = float(base_salary)
        self.sales = float(sales)
        self.commission_rate = float(commission_rate)

    def calculate_pay(self):
        pay = self.base_salary + (self.sales * self.commission_rate)
        return round(pay, 2)


def generate_payroll_summary(employees):
    if not isinstance(employees, list):
        raise TypeError("Employees ek list honi chahiye")

    records = []
    total_payout = 0.0

    for emp in employees:
        if not isinstance(emp, Employee):
            raise TypeError("List ke saare items Employee subclass hone chahiye")

        pay = emp.calculate_pay()
        total_payout += pay
        records.append({
            "emp_id": emp.emp_id,
            "name": emp.name,
            "type": emp.get_role(),
            "pay": pay
        })

    return {
        "employee_count": len(records),
        "total_payout": round(total_payout, 2),
        "details": records
    }


# Testing
if __name__ == "__main__":
    e1 = SalariedEmployee("E01", "Ali", 3000)
    e2 = HourlyEmployee("E02", "Sara", 20, 45)          # Expected: 950.0
    e3 = CommissionEmployee("E03", "Bilal", 500, 4000, 0.10)  # Expected: 900.0

    staff = [e1, e2, e3]
    summary = generate_payroll_summary(staff)

    print("Total Payout:", summary["total_payout"])
    for item in summary["details"]:
        print(item)
```
