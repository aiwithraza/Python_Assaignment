## Task 03 | Recursive JSON Path Explorer

```python
def find_paths(data, target):
    results = []

    def is_exact_match(val, tgt):
        # Python mein True == 1 hota hai, isliye bool ko int se alag rakhna hai
        if type(val) == bool or type(tgt) == bool:
            return type(val) == type(tgt) and val == tgt
        return val == tgt

    def explore(current, current_path):
        # Pehle check karein ke current item hi target to nahi
        if is_exact_match(current, target):
            results.append(current_path)

        # Agar dictionary hai to har key-value ke andar jayein
        if isinstance(current, dict):
            for key, value in current.items():
                if current_path == "":
                    new_path = str(key)
                else:
                    new_path = current_path + "." + str(key)
                explore(value, new_path)

        # Agar list hai to har index ke andar jayein
        elif isinstance(current, list):
            for i in range(len(current)):
                new_path = current_path + "[" + str(i) + "]"
                explore(current[i], new_path)

    explore(data, "")
    return results


# Testing
sample_data = {"a": [{"x": 5}, 5], "b": {"y": 5}}
print(find_paths(sample_data, 5))

# Bool edge case test (True ko 1 nahi samajhna chahiye)
bool_test = {"flag": True, "count": 1, "items": [1, False, True]}
print(find_paths(bool_test, 1))
print(find_paths(bool_test, True))
```
