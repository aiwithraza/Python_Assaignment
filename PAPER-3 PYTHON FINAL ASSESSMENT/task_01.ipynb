## Task 01 | Nested Transaction Analyzer

```python
def analyze_transactions(transactions):
    seen_ids = set()
    customers = {}

    for tx in transactions:
        tx_id = tx["id"]

        if tx_id in seen_ids:
            continue
        seen_ids.add(tx_id)

        customer = tx["customer"]
        
        if customer not in customers:
            customers[customer] = {"total": 0, "categories": {}}

        if tx["status"] == "success":
            amount = tx["amount"]
            category = tx["category"]
            
            customers[customer]["total"] += amount
            customers[customer]["categories"][category] = (
                customers[customer]["categories"].get(category, 0) + amount
            )

    report = [
        {
            "customer": cust,
            "total": data["total"],
            "top_category": max(data["categories"], key=data["categories"].get)
            if data["categories"]
            else None,
        }
        for cust, data in customers.items()
    ]

    report.sort(key=lambda x: (-x["total"], x["customer"]))

    for idx, item in enumerate(report, start=1):
        item["rank"] = idx

    return report

sample_data = [
    {"id": 1, "customer": "Ali", "category": "Food", "amount": 500, "status": "success"},
    {"id": 2, "customer": "Ali", "category": "Travel", "amount": 200, "status": "failed"},
    {"id": 3, "customer": "Sara", "category": "Food", "amount": 700, "status": "success"},
    {"id": 3, "customer": "Sara", "category": "Food", "amount": 700, "status": "success"}, # Duplicate ID
    {"id": 4, "customer": "Zain", "category": "Shopping", "amount": 300, "status": "failed"}, # Only failed
]

result = analyze_transactions(sample_data)
for row in result:
    print(row)
```
