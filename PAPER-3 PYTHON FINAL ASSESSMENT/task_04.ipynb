## Task 04 | Dependency Execution Planner

```python
def plan_execution(tasks, dependencies):
    # Input validation
    if not isinstance(tasks, (list, set, tuple)) or not isinstance(dependencies, (list, set, tuple)):
        raise TypeError("Tasks aur dependencies collections honi chahiye")

    # Duplicate tasks hatane ke liye unique set banayein
    unique_tasks = set(tasks)

    # Graph aur prerequisite count initialize karein
    dependents = {}
    prereq_count = {}

    for t in unique_tasks:
        dependents[t] = []
        prereq_count[t] = 0

    # Dependencies check aur build karein
    for pair in dependencies:
        if len(pair) != 2:
            raise ValueError("Har dependency pair mein 2 tasks hone chahiye")

        before, after = pair

        # Unknown task names reject karein
        if before not in unique_tasks:
            raise ValueError(f"Unknown task naam mila: '{before}'")
        if after not in unique_tasks:
            raise ValueError(f"Unknown task naam mila: '{after}'")

        # Agar duplicate dependency pair na ho to add karein
        if after not in dependents[before]:
            dependents[before].append(after)
            prereq_count[after] += 1

    # Jin tasks ki koi dependency nahi (count == 0), unhein ready list mein daalein
    available = []
    for t in unique_tasks:
        if prereq_count[t] == 0:
            available.append(t)

    execution_order = []

    while len(available) > 0:
        # Alphabetically sabse pehla task select karein
        available.sort()
        current = available.pop(0)
        execution_order.append(current)

        # Is task par depend karne wale tasks ka count kam karein
        for next_task in dependents[current]:
            prereq_count[next_task] -= 1
            if prereq_count[next_task] == 0:
                available.append(next_task)

    # Agar saare tasks complete nahi hue to cycle maujood hai
    if len(execution_order) != len(unique_tasks):
        raise ValueError("Cycle detected: tasks mein circular dependency maujood hai")

    return execution_order


# Testing
tasks_sample = ["build", "test", "deploy"]
deps_sample = [("build", "test"), ("test", "deploy")]
print(plan_execution(tasks_sample, deps_sample))

# Alphabetical tie-break test
print(plan_execution(["c", "a", "b"], []))  # Output: ['a', 'b', 'c']
```
