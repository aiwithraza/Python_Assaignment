## Task 02 | Longest Valid Window

```python
def longest_unique_window(text):
    # Agar input string nahi hai to error raise karo
    if type(text) != str:
        raise TypeError("Input string honi chahiye")

    # Empty string ka edge case
    if len(text) == 0:
        return ("", 0, 0)

    seen = {}
    left = 0
    max_length = 0
    start_index = 0

    for right in range(len(text)):
        ch = text[right]

        # Agar character pehle aa chuka hai aur current window ke andar hai
        if ch in seen and seen[ch] >= left:
            left = seen[ch] + 1

        # Character ki latest position save 
        seen[ch] = right

        # Current window ki length check
        current_length = right - left + 1
        if current_length > max_length:
            max_length = current_length
            start_index = left

    result_str = text[start_index : start_index + max_length]
    return (result_str, start_index, max_length)


# Testing
print(longest_unique_window("abcabcbb"))
print(longest_unique_window(""))
print(longest_unique_window("pwwkew"))
print(longest_unique_window("a b!a"))
print(longest_unique_window("Haider"))
```
